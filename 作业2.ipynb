```python
import cv2
import numpy as np
import PIL
import matplotlib.pyplot as plt
img = cv2.imread("./cv.png")
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)#为了在plt显示时不变色，进行BGR转RGB
Arr = np.float32([[1,0,100],[0,1,150]])
img0 = cv2.warpAffine(img,Arr,(img.shape[1],img.shape[0]))#
img1= cv2.resize(img,(1024,768))#按像素缩小
img2 = cv2.resize(img,None,fx = 0.6,fy = 0.6)#按比例缩小
rows, cols, channel = img.shape
Arr = cv2.getRotationMatrix2D((cols/2, rows/2), 60, 1)
img3 = cv2.flip(img, 1)#水平
img4 = cv2.flip(img, 0)#垂直
img5 = cv2.flip(img, -1)#垂直+水平翻
img6 = cv2.warpAffine(img,Arr,(img.shape[1],img.shape[0]))#旋转
img8 = cv2.resize(img,None,fx = 0.4,fy = 0.4)
img7 = img.copy()
for i in range(img8.shape[1]):#缩小放左上角
    for j in range(img8.shape[0]):
        img7[j][i] = img8[j][i]
titles=['src','img0','img1','img2','img3','img4','img5','img6','img7']
images =[img,img0,img1,img2,img3,img4,img5,img6,img7]
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.xticks([])
    plt.yticks([])
plt.show()
img = cv2.cvtColor(img,cv2.COLOR_RGB2BGR)
img1 = cv2.cvtColor(img1,cv2.COLOR_RGB2BGR)
img2 = cv2.cvtColor(img2,cv2.COLOR_RGB2BGR)
cv2.imshow('src',img)
cv2.imshow('small 1',img1)
cv2.imshow('small 2',img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

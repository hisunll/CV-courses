```python

import cv2
from PIL import ImageFont, ImageDraw,Image
import numpy
```

```python
img = cv2.imread('cv.png',flags=1) #读取
cv2.namedWindow("image", cv2.WINDOW_NORMAL)
cv2.imshow('image',img) #显示图片
fontpath = "./simsun.ttc"
font = ImageFont.truetype(fontpath,32) #字体文件
img_pil = Image.fromarray(img)
draw = ImageDraw.Draw(img_pil)
draw.text((20,20),"20123112 孙梁梁",font = font,fill = (255,0,0,0)) #加文字
img = numpy.array(img_pil) #转array
cv2.namedWindow("image2", cv2.WINDOW_NORMAL)
cv2.imshow('image2',img)
cv2.imwrite('result.png', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
video_test = cv2.VideoCapture('./Waymo.mp4') #读视频
cv2.namedWindow("video", cv2.WINDOW_NORMAL)
while 1: #逐帧读和显示
    ret, frame = video_test.read()
    if frame is None: 
        cv2.waitKey(0)
        break
    cv2.imshow('video',frame)
    if cv2.waitKey(25) & 0xFF == 27: break
cv2.waitKey(0)
video_test.release()
cv2.destoryAllwindows()
```
